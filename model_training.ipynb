```python
from ultralytics import YOLO
from google.colab import files
from IPython.display import display
from PIL import Image

model = YOLO("/content/runs/detect/train/weights/best.pt")    # use train2 later if you train again

uploaded = files.upload()                                     # a button appears: pick your photos

for name in uploaded:
    result = model.predict(name, imgsz=512, conf=0.25, save=True)[0]
    for box in result.boxes:
        print(name, "->", result.names[int(box.cls)], round(float(box.conf), 2))
    display(Image.open(result.save_dir + "/" + name))         # shows the picture with the box
```
